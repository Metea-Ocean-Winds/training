Performing basic calculations in Python

```python
wind_speed = 12.5  # in m/s
blade_length = 50  # in meters
rotor_swept_area = 3.14 * (blade_length ** 2)  # Area of the rotor circle

efficiency = 0.4  # Assumed efficiency of wind turbine

turbine_power = 0.5  * 1.225 * rotor_swept_area * (wind_speed ** 3) * efficiency


print("Rotor Swept Area:", rotor_swept_area, "m^2")
print("Estimated Power Output:", turbine_power/2, "W")
```

### Section 3: Control Structures
Using if-else statements for wind turbine status

```python
wind_speed = float(input("Enter wind speed (m/s): "))

if wind_speed < 3:
    print("Wind speed too low, turbine is not operational.")
elif 3 <= wind_speed <= 25:
    print("Wind speed optimal, turbine is operational.")
else:
    print("Wind speed too high, turbine shut down for safety.")
```

### Section 4: Loops
Simulating wind speed variations over time


```python
wind_speeds = [3, 5, 8, 12, 15, 20, 22, 26]

for speed in wind_speeds:
    if speed > 25:
        print(f"Wind speed {speed} m/s: Turbine shut down.")
    else:
        print(f"Wind speed {speed} m/s: Turbine operational.")
    print('fin if-else')
        
print('fin loop')
```

### Section 5: Functions
Defining a function to calculate turbine power

```python
def calculate_power(wind_speed, blade_length, efficiency=0.5):
    """
    Este es un comentario de la funcion y esto es lo que hay que hacer
    """
    print(wind_speed, blade_length, efficiency)
    rotor_swept_area = 3.14 * (blade_length ** 2)
    power = 0.5 * 1.225 * rotor_swept_area * (wind_speed ** 3) * efficiency    
    return power
```

```python
help(calculate_power)
```

```python
# Example usage
bl = 50  # meters
ws = 10  # m/s
eff = 0.88
turbine_power = calculate_power(blade_length=bl, efficiency=eff, wind_speed=ws)
print(f"Turbine Power Output at {wind_speed} m/s: {turbine_power:.2f} W")
```

```python
wind_farm_data = {
    "Turbine_1": 10.5,
    "Turbine_2": 12.3,
    "Turbine_3": 8.7,
    "Turbine_4": 14.2
}
```

```python
# Iterating through dictionary
total_power = 0
for turbine, speed in wind_farm_data.items():
    power = calculate_power(speed, blade_length)
    print(f"{turbine} - Wind Speed: {speed} m/s - Power Output: {power:.2f} W")
    total_power += power

print(f"Total Wind Farm Power Output: {total_power:.2f} W")
```

### Section 7: Handling Errors with Try-Except
Handling invalid user input

```python
try:
    user_speed = float(input("Enter wind speed (m/s) for power calculation: "))
    user_power = calculate_power(user_speed, blade_length)
    print(f"Estimated Power Output: {user_power:.2f} W")
except ValueError:
    print("Invalid input! Please enter a numeric value.")
```

### Section 8: File Handling
writing wind farm data to a file

```python
with open("wind_farm_data.txt", "w") as file:
    for turbine, speed in wind_farm_data.items():
        power = calculate_power(speed, blade_length)
        file.write(f"{turbine}, Wind Speed: {speed} m/s, Power Output: {power:.2f} W\n")
print("Wind farm data has been saved to wind_farm_data.txt")
```

### Section 9: Reading from a File
Reading and displaying the saved wind farm data

```python
with open("wind_farm_data.txt", "r") as file:
    print("\nReading saved wind farm data:")
    for line in file:
        print(line.strip())
```
