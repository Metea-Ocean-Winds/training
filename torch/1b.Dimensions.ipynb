```python
import torch

# Create sample tensors for examples
x = torch.tensor([1, 2, 3])           # shape: (3,)
y = torch.tensor([[1, 2], [3, 4]])    # shape: (2, 2)
z = torch.randn(2, 1, 3, 1)           # shape: (2, 1, 3, 1)

print("Original shapes:")
print(f"x: {x.shape},\ny: {y.shape},\nz: {z.shape}")
```

#### Adding dimensions with unsqueeze()

```python
print("unsqueeze() - Add dimension at specific position\n")
print("Original shape:")
print(f"x: {x.shape}")

print("\nUnsqueeze examples:")
x_unsqueezed = x.unsqueeze(0)         # shape: (1, 3) - add batch dim
print(f"x.unsqueeze(0): {x_unsqueezed.shape}")
print(x_unsqueezed)

x_unsqueezed2 = x.unsqueeze(1)        # shape: (3, 1) - add column dim
print(f"\nx.unsqueeze(1): {x_unsqueezed2.shape}")
print(x_unsqueezed2)

x_unsqueezed3 = x.unsqueeze(-1)       # shape: (3, 1) - add at end
print(f"\nx.unsqueeze(-1): {x_unsqueezed3.shape}")
print(x_unsqueezed3)
```

#### Alternative: indexing with None

```python

x_none = x[None, :]   # shape: (1, 3) - same as unsqueeze(0)
x_none2 = x[:, None]  # shape: (3, 1) - same as unsqueeze(1)

print("Original shape:")
print(f"x: {x.shape}")
print("\nindexing with None examples:")
print(f"x[None, :]: {x_none.shape}")
print(f"x[:, None]: {x_unsqueezed2.shape}")
```

#### Removing dimensions with squeeze()

```python

print("Original shape:")
print(f"z: {z.shape}")

print("\nsqueeze() - Remove dimensions of size 1\n")
z_squeezed = z.squeeze()              # shape: (2, 3) - removes all size-1 dims
z_squeezed_dim = z.squeeze(1)         # shape: (2, 3, 1) - removes only dim 1
z_squeezed_specific = z.squeeze(-1)   # shape: (2, 1, 3) - removes last dim
z_squeezed_dim2 = z.squeeze(2)        # shape: (2, 1, 3, 1) - dim 2 is not a size-1 dim
print("Squeeze examples:")
print(f"z.squeeze(): {z_squeezed.shape}")
print(f"z.squeeze(1): {z_squeezed_dim.shape}")
print(f"z.squeeze(-1): {z_squeezed_specific.shape}")
print(f"z.squeeze(2): {z_squeezed_dim2.shape}")
```

#### Expanding and Broadcasting Tensors

```python

print("Original shape:")
print(f"x: {x.shape}")
print(x)

print("\nexpand() - Expand tensor to new shape (doesn't copy data)")
print("Note: can only expand size-1 dimensions or add new dimensions\n")

x_expanded = x.unsqueeze(0).expand(4, 3)  # shape: (4, 3)
print(f"x.unsqueeze(0).expand(4, 3) : {x_expanded.shape}")
print(x_expanded)

x_expanded2 = x.expand(4,3)  # shape: (4, 3)
print(f"\nx.expand(4,3): {x_expanded.shape}")
print(x_expanded2)
```

```python

print("a kind of reshaping...")
print("x.unsqueeze(1).expand(3, 4):")
x.unsqueeze(1).expand(3, 4)
```

```python

print("expand_as() - Expand to match another tensor's shape")
target = torch.zeros(4, 3)
x_expand_as = x.unsqueeze(0).expand_as(target)  # shape: (4, 3)
print(f"target: {target.shape}")
print(target)
print(f"x.unsqueeze(0).expand_as(target): {x_expand_as.shape}")
print(x_expand_as)
```

```python

print("repeat() - Actually repeat the data (copies memory)\n")

x_repeated = x.repeat(2, 1)           # shape: (6,) - repeat 2x along dim 0

print("Original shape:")
print(f"x: {x.shape}")
print(x)
print(f"\nx.repeat(2,1): {x_repeated.shape} (repeat 2x along dim 0, the batch dim)")
print(x_repeated)

y_repeated = y.repeat(2, 3)           # shape: (4, 6) - repeat 2x dim0, 3x dim1
print("\nOriginal shape:")
print(f"y: {y.shape}")
print(y)
print(f"\ny.repeat(2, 3): {y_repeated.shape} (repeat 2x dim0, 3x dim1)")
print(y_repeated)
```

#### Reshaping

```python

print("reshape() - Change shape (total elements must match)\n")

x_reshaped = x.reshape(1, 3)          # shape: (1, 3)
print("Original shape:")
print(f"x: {x.shape}")
print(x)
print(f"\nReshape examples:")
print(f"x.reshape(1, 3): {x_reshaped.shape}")
print(x_reshaped)

print("\nreshape() / view() - Change shape (total elements must match)\n")
print("Original shape:")
print(f"y: {y.shape}")
print(y)
y_reshaped = y.reshape(4)             # shape: (4,) - flatten
y_viewed = y.view(-1)                 # shape: (4,) - -1 means infer dimension
print(f"\ny.reshape(4): {y_reshaped.shape}")
print(y_reshaped)
print(f"\ny.view(-1): {y_viewed.shape}")
print(y_viewed)
```

```python

print("flatten() - Flatten from start_dim to end_dim")
batch_data = torch.randn(32, 28, 28)  # batch of images
print(f"batch_data: {batch_data.shape}")
print(batch_data)

flattened = batch_data.flatten(1)     # shape: (32, 784) - keep batch dim
print(f"flatten(1): {flattened.shape}")
print(flattened)
```

```python
# ========== PERMUTING DIMENSIONS ==========

# permute() - Rearrange dimensions
img = torch.randn(3, 224, 224)        # shape: (channels, height, width)
img_hwc = img.permute(1, 2, 0)        # shape: (height, width, channels)

print(f"\nPermute: {img.shape} -> {img_hwc.shape}")

# transpose() - Swap two dimensions
img_transposed = img.transpose(1, 2)   # shape: (3, 224, 224) -> (3, 224, 224)
matrix_t = y.transpose(0, 1)          # shape: (2, 2) -> (2, 2) - matrix transpose
matrix_t2 = y.T                       # shorthand for 2D transpose

# ========== USEFUL COMBINATIONS ==========

# Common pattern: Add batch dimension
single_sample = torch.randn(3, 224, 224)
batched = single_sample.unsqueeze(0)   # shape: (1, 3, 224, 224)

# Common pattern: Flatten except batch dimension
batch_imgs = torch.randn(32, 3, 28, 28)
batch_flat = batch_imgs.view(32, -1)   # shape: (32, 2352)

# Common pattern: Match dimensions for broadcasting
a = torch.randn(32, 10)               # shape: (32, 10)
b = torch.randn(10)                   # shape: (10,)
b_matched = b.unsqueeze(0)            # shape: (1, 10) - now broadcasts with a

print(f"\nCommon patterns:")
print(f"Add batch dim: {single_sample.shape} -> {batched.shape}")
print(f"Flatten batch: {batch_imgs.shape} -> {batch_flat.shape}")
print(f"Broadcasting: {b.shape} -> {b_matched.shape}")

# ========== DIMENSION CHECKING ==========

# Get number of dimensions
print(f"\nDimension info:")
print(f"x.dim(): {x.dim()}")          # number of dimensions
print(f"x.ndim: {x.ndim}")            # same as dim()
print(f"x.size(): {x.size()}")        # same as x.shape
print(f"x.numel(): {x.numel()}")      # total number of elements
```

Key differences to remember:

`view()` vs `reshape()`: `view()` requires contiguous memory, reshape() may copy data<br>
expand() vs repeat(): expand() doesn't copy data (memory efficient), repeat() actually copies<br>
squeeze() vs unsqueeze(): Remove vs add dimensions of size 1<br>
permute() vs transpose(): Rearrange all dimensions vs swap two specific dimensions


Common use cases:

Adding batch dimension: tensor.unsqueeze(0)<br>
Flattening for linear layers: tensor.view(batch_size, -1)<br>
Channel format conversion: tensor.permute(0, 2, 3, 1) (NCHW → NHWC)<br>
Broadcasting preparation: tensor.unsqueeze(-1) or tensor[:, None]<br>

```python
# Example: expand(batch_size, -1, -1)

# Starting tensor (2D)
img = torch.randn(224, 224)  # shape: (224, 224) - single image
print(f"Original: {img.shape}")

# Add a dimension first (common pattern)
img_with_batch = img.unsqueeze(0)  # shape: (1, 224, 224)
print(f"After unsqueeze(0): {img_with_batch.shape}")

# Now expand to batch_size
batch_size = 8
expanded = img_with_batch.expand(batch_size, -1, -1)  # shape: (8, 224, 224)
print(f"After expand(8, -1, -1): {expanded.shape}")

# What the -1 means:
# -1 means "keep this dimension as is"
# So expand(8, -1, -1) means:
# - Make dimension 0 have size 8
# - Keep dimension 1 the same (224)  
# - Keep dimension 2 the same (224)
```

#### Key advantages:
1. `Memory efficiency`: expand() doesn't copy data, just creates views
2. `Broadcasting compatibility`: Makes tensors broadcastable for element-wise operations
3. `Parallel processing`: Same operation applied to multiple samples simultaneously
4. `Consistent initialization`: Ensures all copies start from identical state

#### Real-world scenarios:
1. `Computer Vision`: Applying same data augmentation to batch
2. `NLP`: Using same attention pattern across batch sequences
3. `Reinforcement Learning`: Running multiple parallel environments
4. `Scientific Computing`: Monte Carlo simulations with identical setups
5. `Deep Learning`: Ensemble training with shared initialization
